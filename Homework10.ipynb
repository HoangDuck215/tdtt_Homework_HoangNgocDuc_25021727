```python
#Work2
def merge(lsta, lstb):
    if len(lsta) == 0:
        return lstb
    elif len(lstb) ==0:
        return lsta
    if lsta[0] < lstb[0]:
        return [lsta[0]] + merge(lsta[1::], lstb)
    else:
        return [lstb[0]] + merge(lstb[1::], lsta)
def flatten(matrix):
    if len(matrix) ==1:
        return matrix[0]
    mid = len(matrix) //2
    top_half = matrix[:mid]
    top_down = matrix[mid:]
    merged_top = flatten(top_half)
    merged_down = flatten(top_down)
    return merge(merged_top, merged_down)
def nhap_ma_tran():
    matrix = []
    while True:
        try:
            line = input()
            if line.strip() == "":
                break
            row = list(map(int, line.split()))
            matrix.append(row)
        except:
            break
    return matrix
matrix = nhap_ma_tran()
k = int(input())
ans = flatten(matrix)
print(ans[k-1])
```

```python
#Work5
def subset(lst1, lst2, i, j):
    if i == len(lst1):
        return "YES"
    elif j == len(lst2):
        return "NO"
    if lst1[i] == lst2[j]:
        return subset(lst1, lst2, i+1, j+1)
    else:
        return subset(lst1, lst2, i-i,j-i+1)
lst1 = list(map(int, input().split()))
lst2 = list(map(int, input().split()))
print(subset(lst1, lst2, 0, 0))
```

```python
#Work6
def next(lst, i):
    if i < 0:
        return lst[::-1]
    if lst[i]>lst[i-1]:
        for j in range(len(lst)-1,i-1, -1):
            if lst[j] > lst[i-1]:
                lst[i-1], lst[j] = lst[j], lst[i-1]
                break
        return lst[:i]+lst[i:][::-1]
    else:
        return next(lst, i-1)
n = int(input())
lst = list(map(int, input().split()))
print(*next(lst, n-1))
```

```python
#Work7
def CanPlaceFlowers(flowerbed, k, i=0):
    if k ==0:
        return "True"
    elif i == len(flowerbed):
        return "False"
    if flowerbed[i] == 1:
        return CanPlaceFlowers(flowerbed, k, i+2)
    else:
        if (i==0 or flowerbed[i-1]==0) and (i==len(flowerbed)-1 or flowerbed[i+1]==0):
            return CanPlaceFlowers(flowerbed, k-1, i+1)
flowerbed = list(map(int, input().split()))
k = int(input())
print(CanPlaceFlowers(flowerbed, k))
```

```python
#Work8
def is_path_crossing(moves, a, b, lst,i=0):
    if i == len(moves):
        return False
    if moves[i] == "R":
        a = a+1
    if moves[i] == "L":
        a = a-1
    if moves[i] =="U":
        b = b+1
    if moves[i] == "D":
        b= b-1
    now = (a,b)
    if now in lst:
        return "True"
    lst.append(now)
    return is_path_crossing(moves, a, b, lst,i+1)

s = input()
moves = list(s)
lst = [(0,0)]
print(is_path_crossing(moves, 0, 0, lst))
```

```python
#Work9
def Median(nums):
    if nums[0] > nums[1]:
        nums[0], nums[1] = nums[1],nums[0]
    if nums[2] > nums[3]:
        nums[2],nums[3] = nums[3],nums[2]
    if nums[0] > nums[2]:
        nums[0],nums[2] = nums[2],nums[0]
        nums[1],nums[3] = nums[3],nums[1]
    if nums[1] > nums[4]:
        nums[1],nums[4] = nums[4], nums[1]
    if nums[1] > nums[2]:
        nums[1],nums[2] = nums[2],nums[1]
    if nums[2] > nums[4]:
        nums[2],nums[4] = nums[4],nums[2]
    return nums[2]
nums = list(map(int, input().split()))
print(Median(nums))
```
